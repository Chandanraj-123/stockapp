```python
from pathlib import Path
import pandas as pd
from tradingview_screener import Query

# --------------------------------------------------
# Create output folder
# --------------------------------------------------
Path("data").mkdir(exist_ok=True)

# --------------------------------------------------
# Fetch data
# --------------------------------------------------
count, df = (
    Query()
    .set_markets("india")
    .select(
        "ticker",
        "name",
        "exchange",
        "close",
        "change",
    )
    .limit(10000)
    .get_scanner_data()
)

print(f"TradingView Count : {count}")

# Remove duplicate columns
df = df.loc[:, ~df.columns.duplicated()]

# Keep required columns
df = df[["ticker", "name", "exchange", "close", "change"]]

# Split
nse_df = (
    df[df["exchange"] == "NSE"]
    .sort_values("ticker")
    .reset_index(drop=True)
)

bse_df = (
    df[df["exchange"] == "BSE"]
    .sort_values("ticker")
    .reset_index(drop=True)
)

# --------------------------------------------------
# Save as Excel
# --------------------------------------------------
nse_file = "data/nse.xlsx"
bse_file = "data/bse.xlsx"

nse_df.to_excel(nse_file, index=False, engine="openpyxl")
bse_df.to_excel(bse_file, index=False, engine="openpyxl")

print(f"NSE Stocks : {len(nse_df)}")
print(f"BSE Stocks : {len(bse_df)}")
print(f"Total      : {len(nse_df) + len(bse_df)}")

print(f"\nSaved:")
print(f"  {nse_file}")
print(f"  {bse_file}")
```
